    - From the definition in `numpy`
    - Using an analytic series expansion: $\pi = 4 \sum_{k=0}^\infty \frac{(-1)^{k}}{2k + 1}$

        1. Sample $n$ points uniformly on the unit square using `numpy.random.uniform`

        3. Compare the fraction $\frac{n_{<r}}{n}$ to the ratio of the area of a quadrant of a disc to the area of the square to derive an estimate of $\pi$

```python
def r_fun(x,y):
    r = np.sqrt(x**2 + y**2)
    return r
```

```python
bucket_in = 0
bucket_out = 0

t = 1000000

for i in range(t):
    n_x = np.random.uniform(0,1)
    n_y = np.random.uniform(0,1)
    
    r = r_fun(n_x,n_y)
    
    if r <=1: 
        bucket_in +=1
    else:
        bucket_out +=1


print("Inside the Bucket:", bucket_in)
print("Outside the Bucket:", bucket_out)

ratio = 4*bucket_in/t

print("myPi",ratio)
```

```python
bucket_in = 0
bucket_out = 0

t = 10000

for i in range(t):
    n_x = np.random.uniform(0,1)
    n_y = np.random.uniform(0,1)
    
    r = r_fun(n_x,n_y)
    
    if r <=1: 
        bucket_in +=1
    else:
        bucket_out +=1


print("Inside the Bucket:", bucket_in)
print("Outside the Bucket:", bucket_out)

ratio = 4*bucket_in/t

print("myPi",ratio)
```

```python
# series expansion
import matplotlib.pyplot as plt
num_terms = 100
series_pi = 4 * np.cumsum([(-1)**k / (2*k + 1) for k in range(num_terms)])
plt.plot(series_pi, label='Series pi')
ax = plt.gca()
ax.axhline(np.pi, c='r', ls='--', label='np.pi')
plt.legend()
plt.xlabel("Number of terms")
plt.ylabel("Estimate")
plt.show()
```
